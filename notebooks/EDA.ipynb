# 📊 Nexalyze — Exploratory Data Analysis

Key insights across customers, products, transactions, inventory, and support data.

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from pathlib import Path

pr = Path().resolve().parent / 'data' / 'processed'
read = lambda n: pd.read_csv(pr / f'{n}.csv', low_memory=False)

customers = read('customers')
products  = read('products')
txn       = read('transactions')
tickets   = read('support_tickets')
reviews   = read('customer_reviews')

print('Datasets loaded!')
for name, df in [('customers',customers),('products',products),('transactions',txn)]:
    print(f'{name}: {len(df):,} rows')
```

```python
# Customer Segment Distribution
fig = px.pie(customers, names='segment', title='Customer Segments')
fig.show()
```

```python
# Revenue by category
txn['transaction_date'] = pd.to_datetime(txn['transaction_date'])
prod_cat = products[['product_id','category']]
txn_cat = txn.merge(prod_cat, on='product_id', how='left')
rev_cat = txn_cat.groupby('category')['total_amount'].sum().reset_index().sort_values('total_amount', ascending=False)
fig = px.bar(rev_cat, x='category', y='total_amount', title='Revenue by Category', color='total_amount', color_continuous_scale='Blues')
fig.show()
```

```python
# Monthly revenue trend
txn['month_year'] = txn['transaction_date'].dt.to_period('M').astype(str)
monthly = txn.groupby('month_year')['total_amount'].sum().reset_index()
fig = px.line(monthly, x='month_year', y='total_amount', title='Monthly Revenue Trend', markers=True)
fig.show()
```

```python
# Ticket category distribution
cat_counts = tickets['category'].value_counts().reset_index()
cat_counts.columns = ['category','count']
fig = px.bar(cat_counts, x='count', y='category', orientation='h', title='Support Ticket Categories', color='count', color_continuous_scale='Reds')
fig.show()
```

```python
# Review rating distribution
rating_dist = reviews['rating'].value_counts().sort_index().reset_index()
rating_dist.columns = ['rating','count']
fig = px.bar(rating_dist, x='rating', y='count', title='Review Rating Distribution', color='count', color_continuous_scale='Greens')
fig.show()
```

```python
# Channel split
channel = txn['channel'].value_counts().reset_index()
channel.columns = ['channel','count']
fig = px.pie(channel, names='channel', values='count', title='Online vs Offline Transactions')
fig.show()
```
